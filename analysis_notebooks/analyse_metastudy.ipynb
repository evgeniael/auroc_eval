```python
import json
import csv
import pandas as pd
from collections import Counter
```

### Analyse meta study

```python
path = "/auroc_eval/results/meta-analysis/meta analysis-systematic-final.csv"

metadata = pd.read_csv(path)
```

```python
relevant_studies = metadata[metadata["Relevant"] == "Y"]
```

```python
relevant_studies
```

```python
len(relevant_studies)
```

```python
metric_types = list(relevant_studies["Metric type"])
metric_types = [x.lower() for x in metric_types]

all_matches = []

for x in metric_types:
    if "," in x:
        all_matches.extend([y.strip() for y in x.split(",")])
    else:
        all_matches.append(x)

matches_per_type = dict(Counter(all_matches))
matches_per_type
```

```python
matches_per_type = {'auroc': 40,
 'aucpr': 12,
 'ece': 26,
 'brier': 10,
 'aurac': 11,
 'prr': 8,
 'other': 11,
 'correlation': 8}

matches_per_type_proportion = {}

for x in matches_per_type.keys():
    matches_per_type_proportion[x] = matches_per_type[x] / len(relevant_studies)

matches_per_type_proportion
```
